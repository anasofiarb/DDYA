## 1) Descripción del problema **[1 pts]**

Un vivero está analizando el crecimiento de la producción de nuevas plantas a partir de un
proceso de reproducción controlada.
Se ha observado que el número de nuevas plantas producidas en el mes n sigue el siguiente
comportamiento:
• En el mes 0 se produjeron 0 nuevas plantas.
• En el mes 1 se produjo 1 nueva planta.
• A partir del mes 2, el número de nuevas plantas producidas es igual a la suma de las plantas
producidas durante los dos meses anteriores, debido a que las plantas que alcanzan su etapa de
reproducción generan nuevos brotes siguiendo este patrón de crecimiento.
Relación de recurrencia:
T(n) = 0 si n = 0
T(n) = 1 si n = 1
T(n) = T(n-1) + T(n-2) si n ≥ 2
Determinar con memoización cuántas nuevas plantas serán producidas en el mes n.

## 4) Diagramas de flujo **[5 pts]**

Primero llamamos a T(n) se fija si n es 0 o 1. Si sí, devuelve ese número y ya (0 o 1), no hace nada más.
En caso de que no, mira si ese n ya lo había calculado antes, Si ya estaba guardado, lo devuelve.
Si tampoco estaba guardado, se calcula: suma T(n-1) mas T(n-2).
Ese resultado lo guarda para no tener que volver a calcularlo despues.
Y al final devuelve T(n).


### 5) Análisis de complejidad **[15 pts]**
- Justificación:
Si la función se llama a sí misma sin guardar nada, cada vez que calcula T(n) tiene que calcular T(n-1) y T(n-2), y esas dos vuelven a calcular otras dos, y así siguen y siguen. El problema es que termina calculando lo mismo varias veces sin darse cuenta. Por eso sin memoización esto se llama O(2ⁿ).

Con memoización, la función se acuerda de lo que ya calculó antes, guardándolo en un diccionario llamado memo. Si ya sabe el resultado, no lo vuelve a calcular, solo lo busca. Como cada número de 0 hasta n se calcula solo una vez, esto se llama O(n).

- Función T[n]:
T(n) = 0, si n = 0
T(n) = 1, si n = 1
T(n) = T(n-1) + T(n-2), si n ≥ 2

- Mejor y peor caso con gráfica
Peor caso: T(n) = 2·T(n-1) ⟹ O(2ⁿ)
Mejor caso: T(n) = n·O(1) ⟹ O(n)

## 6) Código funcional **[15 pts]**



```python
guardar = {}

def plantas(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    if n in guardar:
        return guardar[n]
    resultado = plantas(n - 1) + plantas(n - 2)
    guardar[n] = resultado
    return resultado


mes = input("Ingresa el NUMERO de mes que quieres consultar: ")

try:
    mes = int(mes)
    if mes < 0:
        print("El mes no puede ser negativo.")
    else:
        resultado = plantas(mes)
        print("El número de nuevas plantas producidas en el mes", mes, "es:", resultado)
except ValueError:
    print("Por favor ingresa un número entero válido.")
```
